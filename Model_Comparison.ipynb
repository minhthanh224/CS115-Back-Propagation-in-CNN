# CNN Model Comparison: 1-Block vs 2-Blocks vs 3-Blocks

This notebook compares the performance and results of three CNN architectures with different numbers of convolutional blocks.


```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.patches import Rectangle
import seaborn as sns

# Set style for better visualizations
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
```

```python
# Calculate number of parameters for each model

def count_conv_params(in_channels, out_channels, kernel_size):
    """Count parameters in a Conv2D layer"""
    weights = in_channels * out_channels * kernel_size * kernel_size
    biases = out_channels
    return weights + biases

def count_dense_params(in_features, out_features):
    """Count parameters in a Dense layer"""
    weights = in_features * out_features
    biases = out_features
    return weights + biases

# Model 1: 1-Block
model1_params = {
    'Conv1': count_conv_params(1, 8, 3),
    'FC1': count_dense_params(8*14*14, 64),
    'FC2': count_dense_params(64, 10)
}
model1_total = sum(model1_params.values())

# Model 2: 2-Blocks
model2_params = {
    'Conv1': count_conv_params(1, 8, 3),
    'Conv2': count_conv_params(8, 16, 3),
    'FC1': count_dense_params(16*7*7, 64),
    'FC2': count_dense_params(64, 10)
}
model2_total = sum(model2_params.values())

# Model 3: 3-Blocks
model3_params = {
    'Conv1': count_conv_params(1, 8, 3),
    'Conv2': count_conv_params(8, 16, 3),
    'Conv3': count_conv_params(16, 32, 3),
    'FC': count_dense_params(32*3*3, 10)
}
model3_total = sum(model3_params.values())

# Create summary table
params_df = pd.DataFrame({
    'Model': ['1-Block', '2-Blocks', '3-Blocks'],
    'Conv Blocks': [1, 2, 3],
    'Total Parameters': [model1_total, model2_total, model3_total],
    'Conv Parameters': [
        model1_params['Conv1'],
        model2_params['Conv1'] + model2_params['Conv2'],
        model3_params['Conv1'] + model3_params['Conv2'] + model3_params['Conv3']
    ],
    'Dense Parameters': [
        model1_params['FC1'] + model1_params['FC2'],
        model2_params['FC1'] + model2_params['FC2'],
        model3_params['FC']
    ]
})

print("\n=== MODEL PARAMETERS COMPARISON ===")
print(params_df.to_string(index=False))

# Visualize parameter distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Bar chart of total parameters
axes[0].bar(params_df['Model'], params_df['Total Parameters'], color=['#3498db', '#e67e22', '#e74c3c'])
axes[0].set_ylabel('Number of Parameters', fontsize=12)
axes[0].set_title('Total Parameters by Model', fontsize=14, fontweight='bold')
axes[0].grid(axis='y', alpha=0.3)

# Add value labels on bars
for i, v in enumerate(params_df['Total Parameters']):
    axes[0].text(i, v + 1000, f'{v:,}', ha='center', va='bottom', fontsize=10, fontweight='bold')

# Stacked bar chart of parameter types
x_pos = np.arange(len(params_df['Model']))
axes[1].bar(x_pos, params_df['Conv Parameters'], label='Conv Parameters', color='#3498db')
axes[1].bar(x_pos, params_df['Dense Parameters'], bottom=params_df['Conv Parameters'], 
            label='Dense Parameters', color='#e67e22')
axes[1].set_xticks(x_pos)
axes[1].set_xticklabels(params_df['Model'])
axes[1].set_ylabel('Number of Parameters', fontsize=12)
axes[1].set_title('Parameter Distribution: Conv vs Dense', fontsize=14, fontweight='bold')
axes[1].legend()
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Training results from the three models
# These are example results - you should replace with actual training runs

epochs = np.arange(1, 11)

# Model 1: 1-Block (example data)
model1_train_loss = [0.2153, 0.0731, 0.0506, 0.0380, 0.0305, 0.0257, 0.0221, 0.0195, 0.0175, 0.0160]
model1_train_acc = [0.9312, 0.9769, 0.9835, 0.9875, 0.9897, 0.9910, 0.9922, 0.9930, 0.9935, 0.9940]
model1_val_acc = [0.9755, 0.9801, 0.9831, 0.9811, 0.9824, 0.9820, 0.9825, 0.9830, 0.9828, 0.9831]

# Model 2: 2-Blocks (example data)
model2_train_loss = [0.2153, 0.0731, 0.0506, 0.0380, 0.0305, 0.0257, 0.0221, 0.0195, 0.0175, 0.0160]
model2_train_acc = [0.9312, 0.9769, 0.9835, 0.9875, 0.9897, 0.9910, 0.9922, 0.9930, 0.9935, 0.9940]
model2_val_acc = [0.9755, 0.9801, 0.9831, 0.9811, 0.9824, 0.9820, 0.9825, 0.9830, 0.9828, 0.9831]

# Model 3: 3-Blocks (example data)
model3_train_loss = [0.2153, 0.0731, 0.0506, 0.0380, 0.0305, 0.0257, 0.0221, 0.0195, 0.0175, 0.0160]
model3_train_acc = [0.9312, 0.9769, 0.9835, 0.9875, 0.9897, 0.9910, 0.9922, 0.9930, 0.9935, 0.9940]
model3_val_acc = [0.9755, 0.9801, 0.9831, 0.9811, 0.9824, 0.9820, 0.9825, 0.9830, 0.9828, 0.9831]

print("Training results loaded successfully!")
print(f"\nFinal Results:")
print(f"1-Block Model  - Train Acc: {model1_train_acc[-1]*100:.2f}%, Val Acc: {model1_val_acc[-1]*100:.2f}%")
print(f"2-Blocks Model - Train Acc: {model2_train_acc[-1]*100:.2f}%, Val Acc: {model2_val_acc[-1]*100:.2f}%")
print(f"3-Blocks Model - Train Acc: {model3_train_acc[-1]*100:.2f}%, Val Acc: {model3_val_acc[-1]*100:.2f}%")
```

```python
plt.figure(figsize=(12, 6))

plt.plot(epochs, model1_train_loss, 'o-', linewidth=2, markersize=8, label='1-Block Model', color='#3498db')
plt.plot(epochs, model2_train_loss, 's-', linewidth=2, markersize=8, label='2-Blocks Model', color='#e67e22')
plt.plot(epochs, model3_train_loss, '^-', linewidth=2, markersize=8, label='3-Blocks Model', color='#e74c3c')

plt.xlabel('Epoch', fontsize=13, fontweight='bold')
plt.ylabel('Training Loss', fontsize=13, fontweight='bold')
plt.title('Training Loss Comparison', fontsize=15, fontweight='bold', pad=20)
plt.legend(fontsize=11, loc='upper right')
plt.grid(True, alpha=0.3)
plt.xticks(epochs)

plt.tight_layout()
plt.show()

print("\n=== FINAL TRAINING LOSS ===")
print(f"1-Block:  {model1_train_loss[-1]:.4f}")
print(f"2-Blocks: {model2_train_loss[-1]:.4f}")
print(f"3-Blocks: {model3_train_loss[-1]:.4f}")
```

```python
plt.figure(figsize=(12, 6))

plt.plot(epochs, np.array(model1_train_acc)*100, 'o-', linewidth=2, markersize=8, 
         label='1-Block Model', color='#3498db')
plt.plot(epochs, np.array(model2_train_acc)*100, 's-', linewidth=2, markersize=8, 
         label='2-Blocks Model', color='#e67e22')
plt.plot(epochs, np.array(model3_train_acc)*100, '^-', linewidth=2, markersize=8, 
         label='3-Blocks Model', color='#e74c3c')

plt.xlabel('Epoch', fontsize=13, fontweight='bold')
plt.ylabel('Training Accuracy (%)', fontsize=13, fontweight='bold')
plt.title('Training Accuracy Comparison', fontsize=15, fontweight='bold', pad=20)
plt.legend(fontsize=11, loc='lower right')
plt.grid(True, alpha=0.3)
plt.xticks(epochs)
plt.ylim([90, 100])

plt.tight_layout()
plt.show()

print("\n=== FINAL TRAINING ACCURACY ===")
print(f"1-Block:  {model1_train_acc[-1]*100:.2f}%")
print(f"2-Blocks: {model2_train_acc[-1]*100:.2f}%")
print(f"3-Blocks: {model3_train_acc[-1]*100:.2f}%")
```

```python
plt.figure(figsize=(12, 6))

plt.plot(epochs, np.array(model1_val_acc)*100, 'o-', linewidth=2, markersize=8, 
         label='1-Block Model', color='#3498db')
plt.plot(epochs, np.array(model2_val_acc)*100, 's-', linewidth=2, markersize=8, 
         label='2-Blocks Model', color='#e67e22')
plt.plot(epochs, np.array(model3_val_acc)*100, '^-', linewidth=2, markersize=8, 
         label='3-Blocks Model', color='#e74c3c')

plt.xlabel('Epoch', fontsize=13, fontweight='bold')
plt.ylabel('Validation Accuracy (%)', fontsize=13, fontweight='bold')
plt.title('Validation Accuracy Comparison', fontsize=15, fontweight='bold', pad=20)
plt.legend(fontsize=11, loc='lower right')
plt.grid(True, alpha=0.3)
plt.xticks(epochs)
plt.ylim([97, 99])

plt.tight_layout()
plt.show()

print("\n=== FINAL VALIDATION ACCURACY ===")
print(f"1-Block:  {model1_val_acc[-1]*100:.2f}%")
print(f"2-Blocks: {model2_val_acc[-1]*100:.2f}%")
print(f"3-Blocks: {model3_val_acc[-1]*100:.2f}%")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Model 1
axes[0].plot(epochs, np.array(model1_train_acc)*100, 'b-', linewidth=2, label='Training', marker='o')
axes[0].plot(epochs, np.array(model1_val_acc)*100, 'orange', linewidth=2, label='Validation', marker='s')
axes[0].set_xlabel('Epoch', fontsize=12)
axes[0].set_ylabel('Accuracy (%)', fontsize=12)
axes[0].set_title('1-Block Model', fontsize=14, fontweight='bold')
axes[0].legend(fontsize=10)
axes[0].grid(True, alpha=0.3)
axes[0].set_ylim([90, 100])

# Model 2
axes[1].plot(epochs, np.array(model2_train_acc)*100, 'b-', linewidth=2, label='Training', marker='o')
axes[1].plot(epochs, np.array(model2_val_acc)*100, 'orange', linewidth=2, label='Validation', marker='s')
axes[1].set_xlabel('Epoch', fontsize=12)
axes[1].set_ylabel('Accuracy (%)', fontsize=12)
axes[1].set_title('2-Blocks Model', fontsize=14, fontweight='bold')
axes[1].legend(fontsize=10)
axes[1].grid(True, alpha=0.3)
axes[1].set_ylim([90, 100])

# Model 3
axes[2].plot(epochs, np.array(model3_train_acc)*100, 'b-', linewidth=2, label='Training', marker='o')
axes[2].plot(epochs, np.array(model3_val_acc)*100, 'orange', linewidth=2, label='Validation', marker='s')
axes[2].set_xlabel('Epoch', fontsize=12)
axes[2].set_ylabel('Accuracy (%)', fontsize=12)
axes[2].set_title('3-Blocks Model', fontsize=14, fontweight='bold')
axes[2].legend(fontsize=10)
axes[2].grid(True, alpha=0.3)
axes[2].set_ylim([90, 100])

plt.suptitle('Training vs Validation Accuracy for All Models', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Calculate overfitting gap (Train Acc - Val Acc)
model1_gap = np.array(model1_train_acc) - np.array(model1_val_acc)
model2_gap = np.array(model2_train_acc) - np.array(model2_val_acc)
model3_gap = np.array(model3_train_acc) - np.array(model3_val_acc)

plt.figure(figsize=(12, 6))

plt.plot(epochs, model1_gap*100, 'o-', linewidth=2, markersize=8, label='1-Block Model', color='#3498db')
plt.plot(epochs, model2_gap*100, 's-', linewidth=2, markersize=8, label='2-Blocks Model', color='#e67e22')
plt.plot(epochs, model3_gap*100, '^-', linewidth=2, markersize=8, label='3-Blocks Model', color='#e74c3c')

plt.axhline(y=0, color='black', linestyle='--', linewidth=1, alpha=0.5)

plt.xlabel('Epoch', fontsize=13, fontweight='bold')
plt.ylabel('Accuracy Gap (%) [Train - Val]', fontsize=13, fontweight='bold')
plt.title('Overfitting Analysis: Training-Validation Gap', fontsize=15, fontweight='bold', pad=20)
plt.legend(fontsize=11, loc='upper left')
plt.grid(True, alpha=0.3)
plt.xticks(epochs)

plt.tight_layout()
plt.show()

print("\n=== OVERFITTING GAP (Final Epoch) ===")
print(f"1-Block:  {model1_gap[-1]*100:.2f}% (Lower is better)")
print(f"2-Blocks: {model2_gap[-1]*100:.2f}% (Lower is better)")
print(f"3-Blocks: {model3_gap[-1]*100:.2f}% (Lower is better)")
print("\nNote: Smaller gap indicates less overfitting")
```

```python
# Create comprehensive summary
summary_df = pd.DataFrame({
    'Model': ['1-Block', '2-Blocks', '3-Blocks'],
    'Conv Blocks': [1, 2, 3],
    'Total Parameters': [model1_total, model2_total, model3_total],
    'Final Train Loss': [model1_train_loss[-1], model2_train_loss[-1], model3_train_loss[-1]],
    'Final Train Acc (%)': [model1_train_acc[-1]*100, model2_train_acc[-1]*100, model3_train_acc[-1]*100],
    'Final Val Acc (%)': [model1_val_acc[-1]*100, model2_val_acc[-1]*100, model3_val_acc[-1]*100],
    'Overfit Gap (%)': [model1_gap[-1]*100, model2_gap[-1]*100, model3_gap[-1]*100]
})

print("\n" + "="*100)
print("FINAL PERFORMANCE SUMMARY")
print("="*100)
print(summary_df.to_string(index=False))
print("="*100)

# Visualize final performance
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

models = ['1-Block', '2-Blocks', '3-Blocks']
colors = ['#3498db', '#e67e22', '#e74c3c']

# Final Training Accuracy
train_accs = [model1_train_acc[-1]*100, model2_train_acc[-1]*100, model3_train_acc[-1]*100]
bars1 = axes[0].bar(models, train_accs, color=colors, edgecolor='black', linewidth=1.5)
axes[0].set_ylabel('Accuracy (%)', fontsize=12, fontweight='bold')
axes[0].set_title('Final Training Accuracy', fontsize=14, fontweight='bold')
axes[0].set_ylim([95, 100])
axes[0].grid(axis='y', alpha=0.3)
for i, v in enumerate(train_accs):
    axes[0].text(i, v + 0.1, f'{v:.2f}%', ha='center', va='bottom', fontsize=11, fontweight='bold')

# Final Validation Accuracy
val_accs = [model1_val_acc[-1]*100, model2_val_acc[-1]*100, model3_val_acc[-1]*100]
bars2 = axes[1].bar(models, val_accs, color=colors, edgecolor='black', linewidth=1.5)
axes[1].set_ylabel('Accuracy (%)', fontsize=12, fontweight='bold')
axes[1].set_title('Final Validation Accuracy', fontsize=14, fontweight='bold')
axes[1].set_ylim([95, 100])
axes[1].grid(axis='y', alpha=0.3)
for i, v in enumerate(val_accs):
    axes[1].text(i, v + 0.1, f'{v:.2f}%', ha='center', va='bottom', fontsize=11, fontweight='bold')

# Overfitting Gap
gaps = [model1_gap[-1]*100, model2_gap[-1]*100, model3_gap[-1]*100]
bars3 = axes[2].bar(models, gaps, color=colors, edgecolor='black', linewidth=1.5)
axes[2].set_ylabel('Gap (%)', fontsize=12, fontweight='bold')
axes[2].set_title('Overfitting Gap (Train - Val)', fontsize=14, fontweight='bold')
axes[2].axhline(y=0, color='black', linestyle='--', linewidth=1, alpha=0.5)
axes[2].grid(axis='y', alpha=0.3)
for i, v in enumerate(gaps):
    axes[2].text(i, v + 0.05, f'{v:.2f}%', ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.suptitle('Final Performance Metrics Comparison', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```
